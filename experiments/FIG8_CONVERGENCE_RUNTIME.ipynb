# SPARC-V2X training convergence and runtime figures

```python
import json, numpy as np
from pathlib import Path
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
ROOT=Path.cwd().parent if Path.cwd().name=="experiments" else Path.cwd()
OUT=ROOT/"results"/"final_figures_v2"
FONT=55; TICK=49; LEG=44; LW=5; MS=16; FIG=(20,18)
log=json.loads((ROOT/"results"/"residual_task_cf_training.json").read_text())
ep=np.array([x["episode"] for x in log]); ret=np.array([x["return"] for x in log]); adv=np.array([x["cf_mean"] for x in log])
ma=np.convolve(ret,np.ones(5)/5,mode="valid"); adv_ma=np.convolve(adv,np.ones(5)/5,mode="valid")
def base(ax,xlab,ylab):
 ax.set_xlabel(xlab,fontsize=FONT); ax.set_ylabel(ylab,fontsize=FONT); ax.tick_params(axis="both",labelsize=TICK,width=2.5,length=10); ax.grid(True,alpha=.28,linewidth=1.5)
def save(fig,stem):
 fig.tight_layout(); fig.savefig(OUT/(stem+".png"),dpi=300,bbox_inches="tight"); fig.savefig(OUT/(stem+".pdf"),bbox_inches="tight"); plt.close(fig)
```

```python
fig,ax=plt.subplots(figsize=FIG); ax.plot(ep,ret,color="#1f77b4",lw=LW); base(ax,"Training episode","Episode completed-value return"); save(fig,"fig8a_convergence_reward")
```

```python
fig,ax=plt.subplots(figsize=FIG); ax.plot(ep[4:],ma,color="#d62728",lw=LW); base(ax,"Training episode","Five-episode moving-average completed-value return"); save(fig,"fig8b_training_return_ma")
```

```python
fig,ax=plt.subplots(figsize=FIG); ax.plot(ep,adv,color="#2ca02c",lw=LW,alpha=.45,label="Raw counterfactual advantage"); ax.plot(ep[4:],adv_ma,color="#1f77b4",lw=LW,label="Five-episode moving average"); ax.axhline(0,color="black",linestyle="--",lw=3,label="Zero"); base(ax,"Training episode","Sampled minus anchor counterfactual advantage"); ax.legend(fontsize=LEG,frameon=False); save(fig,"fig8c_cf_advantage")
```

```python
runtime=json.loads((ROOT/"results"/"final_runtime.json").read_text())["by_density"]; dens=np.array([x["density"] for x in runtime]); mean=np.array([x["mean_ms"] for x in runtime]); p95=np.array([x["p95_ms"] for x in runtime]); fig,ax=plt.subplots(figsize=FIG); ax.plot(dens,mean,color="#1f77b4",marker="o",markersize=MS,lw=LW,label="Mean"); ax.plot(dens,p95,color="#d62728",marker="s",markersize=MS,lw=LW,label="P95"); ax.set_ylim(0,max(10.0,float(p95.max()))*1.08); ax.axhline(10,color="black",linestyle="--",lw=3,label="10 ms control interval"); base(ax,"Vehicle density","Decision latency (ms)"); ax.legend(fontsize=LEG,frameon=False); save(fig,"fig8d_runtime")
```

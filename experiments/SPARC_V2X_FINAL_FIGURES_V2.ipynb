```python
import pandas as pd, json, matplotlib.pyplot as plt
from pathlib import Path
FIGSIZE=(20,18); FONTSIZE=50; LINEWIDTH=6; MARKERS=['o','s','^','D','v','P','X']; COLORS=['#1f77b4','#d62728','#2ca02c','#9467bd','#ff7f0e','#8c564b','#e377c2']; out=Path('results/final_figures_v2'); out.mkdir(exist_ok=True); df=pd.read_csv(out/'figure_data.csv')
train=json.load(open('results/residual_task_cf_training.json')); ep=[r['episode'] for r in train]; reward=[r['return'] for r in train]; signal=[r['cf_mean'] for r in train]
def moving(x,w=5): return [sum(x[max(0,i-w+1):i+1])/len(x[max(0,i-w+1):i+1]) for i in range(len(x))]
```

```python
fig,ax=plt.subplots(figsize=FIGSIZE);ax.plot(ep,reward,color=COLORS[0],lw=LINEWIDTH,alpha=.35);ax.plot(ep,moving(reward),color=COLORS[0],lw=LINEWIDTH);ax.set_xlabel('Episode',fontsize=FONTSIZE);ax.set_ylabel('Completed-value reward',fontsize=FONTSIZE);ax.tick_params(labelsize=FONTSIZE-2);ax.grid(alpha=.25);fig.tight_layout();fig.savefig(out/'fig8a_convergence_reward.png',dpi=180);fig.savefig(out/'fig8a_convergence_reward.pdf');plt.show()
```

```python
loss=[abs(x) for x in signal];fig,ax=plt.subplots(figsize=FIGSIZE);ax.plot(ep,loss,color=COLORS[1],lw=LINEWIDTH,alpha=.35);ax.plot(ep,moving(loss),color=COLORS[1],lw=LINEWIDTH);ax.set_xlabel('Episode',fontsize=FONTSIZE);ax.set_ylabel('Absolute counterfactual learning signal',fontsize=FONTSIZE);ax.tick_params(labelsize=FONTSIZE-2);ax.grid(alpha=.25);fig.tight_layout();fig.savefig(out/'fig8b_convergence_loss_signal.png',dpi=180);fig.savefig(out/'fig8b_convergence_loss_signal.pdf');plt.show()
```

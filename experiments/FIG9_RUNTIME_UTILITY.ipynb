# Figure 9: runtime and utility tradeoff

```python
import json, numpy as np
from pathlib import Path
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
ROOT=Path.cwd().parent if Path.cwd().name=="experiments" else Path.cwd(); OUT=ROOT/"results"/"final_figures_v2"
D=[10,25,50,80,100,125,150,175,200]; labels=["SPARC-V2X","Value-Aware","Minimum-Finish","Task-RL"]; raw=json.loads((ROOT/"results"/"task_aware_full_kpis_10seed.json").read_text())["summary"]; im={"SPARC-V2X":"residual_rl","Value-Aware":"value","Minimum-Finish":"minfinish","Task-RL":"task_rl"}; lookup={(r["method"],r["density"]):r for r in raw}; vals={l:{"R_value":[lookup[(im[l],d)]["R_value"] for d in D],"mean":[lookup[(im[l],d)]["decision_ms_mean"] for d in D],"p95":[lookup[(im[l],d)]["decision_ms_p95"] for d in D]} for l in labels}
styles={"SPARC-V2X":("#1f77b4","o"),"Value-Aware":("#d62728","s"),"Minimum-Finish":("#2ca02c","^"),"Task-RL":("#9467bd","D")}; FONT=55; TICK=49; LEG=40; LW=5; MS=16; FIG=(20,18)
def base(ax,x,y): ax.set_xlabel(x,fontsize=FONT); ax.set_ylabel(y,fontsize=FONT); ax.tick_params(axis="both",labelsize=TICK); ax.grid(True,alpha=.28)
def save(fig,stem): fig.tight_layout(); fig.savefig(OUT/(stem+".png"),dpi=300,bbox_inches="tight"); fig.savefig(OUT/(stem+".pdf"),bbox_inches="tight"); plt.close(fig)
```

```python
fig,ax=plt.subplots(figsize=FIG)
for label in labels:
 c,m=styles[label]; ax.plot(D,vals[label]["mean"],color=c,marker=m,markersize=MS,lw=LW,label=f"{label} mean"); ax.plot(D,vals[label]["p95"],color=c,marker=m,markersize=MS,lw=LW,ls="--",label=f"{label} P95")
ax.axhline(10,color="black",ls=":",lw=3,label="10-ms decision interval"); base(ax,"Vehicle density","Decision latency (ms)"); ax.legend(fontsize=LEG,frameon=False,ncol=2); save(fig,"fig9a_runtime_comparison")
```

```python
fig,ax=plt.subplots(figsize=FIG)
for label in labels:
 c,m=styles[label]; ax.plot(vals[label]["mean"],vals[label]["R_value"],color=c,marker=m,markersize=MS,lw=LW,label=label)
base(ax,"Mean decision latency (ms)",r"Value-weighted completion ratio, $R_{value}$"); ax.legend(fontsize=LEG,frameon=False); save(fig,"fig9b_utility_runtime_tradeoff")
```
